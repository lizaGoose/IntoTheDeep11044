```python
import ikpy.chain
import ikpy.utils.plot as plot_utils

import numpy as np
import time
import math

import ipywidgets as widgets
```

```python
arm = ikpy.chain.Chain.from_urdf_file("arm.urdf", active_links_mask=[False, True, True, True, True])
```

```python
x = 0
y = 0
z = 0
target_position = [ x, y, z]
target_orientation = [0, 0, 0]
```

```python
ik = arm.inverse_kinematics(target_position, target_orientation, orientation_mode="all")
a =  list(map(lambda r:math.degrees(r),ik.tolist()))
verticalHandAngle = a[1]
horizontalHandAngle = a[2]
verticalClawAngle = a[3]
```

```python
import matplotlib.pyplot as plt
fig, ax = plot_utils.init_3d_figure()
fig.set_figheight(9)  
fig.set_figwidth(13)  
arm.plot(ik, ax, target=target_position)
plt.xlim(-20, 20)
plt.ylim(-20, 20)
ax.set_zlim(0, 40)
plt.show()
```
